```python
df = pd.read_csv("social_media_engagement_5000.csv")

print("Dataset loaded successfully.")
print("Dataset Shape:", df.shape)
```

```python
print("First 5 rows of the dataset:")
display(df.head())
```

```python
print("Last 5 rows of the dataset:")
display(df.tail())
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
for column in df.columns:
    print(column)
```

```python
print("Dataset Information:")
df.info()

print("\nData Types:")
display(df.dtypes)
```

```python
df["posted_at"] = pd.to_datetime(
    df["posted_at"],
    dayfirst=True,
    errors="coerce"
)

print("Datatype of posted_at:")
print(df["posted_at"].dtype)

print("\nDate Range:")
print("Minimum Date:", df["posted_at"].min())
print("Maximum Date:", df["posted_at"].max())
```

```python
print("Missing Values in Each Column:")
missing_values = df.isnull().sum()
display(missing_values)

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

display(missing_summary)
```

```python
numeric_missing_columns = [
    "age",
    "likes",
    "comments",
    "shares"
]

for column in numeric_missing_columns:
    median_value = df[column].median()

    print(column, "Median =", median_value)

    df[column] = df[column].fillna(median_value)
```

```python
categorical_missing_columns = [
    "gender",
    "sentiment"
]

for column in categorical_missing_columns:

    mode_value = df[column].mode()[0]

    print(column, "Mode =", mode_value)

    df[column] = df[column].fillna(mode_value)
```

```python
print("Missing Values After Treatment:")
display(df.isnull().sum())

print("\nTotal Missing Values Remaining:")
print(df.isnull().sum().sum())
```

```python
categorical_columns = [
    "gender",
    "country",
    "post_type",
    "post_category",
    "device_type",
    "sentiment"
]

for column in categorical_columns:

    df[column] = (
        df[column]
        .astype(str)
        .str.strip()
        .str.lower()
    )

print("Categorical values standardized successfully.")
```

```python
for column in categorical_columns:

    print("\n", column.upper())
    print(df[column].value_counts())
```

```python
check_columns = [
    "age",
    "likes",
    "comments",
    "shares",
    "watch_time_sec",
    "impression_count",
    "follower_count",
    "engagement_rate"
]

for column in check_columns:

    negative_count = (df[column] < 0).sum()

    print(
        column,
        "- Negative Values:",
        negative_count,
        "- Minimum:",
        df[column].min(),
        "- Maximum:",
        df[column].max()
    )
```

```python
df["sentiment"] = (
    df["sentiment"]
    .str.strip()
    .str.lower()
)

print("Sentiment Distribution:")
display(df["sentiment"].value_counts())
```

```python
df["hashtag_count"] = (
    df["hashtags"]
    .fillna("")
    .astype(str)
    .apply(
        lambda x:
        len([
            hashtag
            for hashtag in x.split()
            if hashtag.startswith("#")
        ])
    )
)

print("Hashtag Count Distribution:")
display(df["hashtag_count"].value_counts().sort_index())

display(
    df[
        ["hashtags", "hashtag_count"]
    ].head(10)
)
```

```python
df["engagement_score"] = (
    df["likes"]
    + df["comments"]
    + df["shares"]
)

display(
    df[
        [
            "likes",
            "comments",
            "shares",
            "engagement_score"
        ]
    ].head()
)
```

```python
df["engagement_rate_calculated"] = (
    df["engagement_score"]
    / df["impression_count"]
)

# Using the recalculated engagement rate
# after missing engagement metrics have been cleaned

df["engagement_rate"] = df["engagement_rate_calculated"]

display(
    df[
        [
            "engagement_score",
            "impression_count",
            "engagement_rate"
        ]
    ].head()
)
```

```python
likes_array = np.array(df["likes"])
comments_array = np.array(df["comments"])
shares_array = np.array(df["shares"])

print("Likes Array:")
print(likes_array[:10])

print("\nAverage Likes using NumPy:")
print(np.mean(likes_array))

print("\nMedian Likes using NumPy:")
print(np.median(likes_array))

print("\nMaximum Likes:")
print(np.max(likes_array))

print("\nMinimum Likes:")
print(np.min(likes_array))
```

```python
print("Summary Statistics for Numerical Columns:")

display(
    df.describe().T
)
```

```python
print("Summary of Categorical Columns:")

display(
    df.describe(include="object").T
)
```

```python
print("Gender Distribution:")
display(df["gender"].value_counts())

print("\nCountry Distribution:")
display(df["country"].value_counts())

print("\nPost Type Distribution:")
display(df["post_type"].value_counts())

print("\nPost Category Distribution:")
display(df["post_category"].value_counts())

print("\nSentiment Distribution:")
display(df["sentiment"].value_counts())

print("\nDevice Distribution:")
display(df["device_type"].value_counts())
```

```python
columns_to_check = [
    "gender",
    "country",
    "post_type",
    "post_category",
    "device_type",
    "sentiment"
]

for column in columns_to_check:

    print("\nColumn:", column)

    print("Unique Values:")
    print(df[column].unique())

    print("Number of Unique Values:")
    print(df[column].nunique())
```

```python
print("Selected Columns:")
display(
    df[
        [
            "post_type",
            "post_category",
            "likes",
            "comments",
            "shares"
        ]
    ].head(10)
)
```

```python
high_engagement_posts = df[
    df["engagement_rate"] >
    df["engagement_rate"].median()
]

print(
    "Number of posts above median engagement:",
    len(high_engagement_posts)
)

display(
    high_engagement_posts.head()
)
```

```python
numeric_df = df.select_dtypes(
    include=np.number
)

correlation_matrix = numeric_df.corr()

display(
    correlation_matrix.round(2)
)
```

```python
avg_likes_post_type = (
    df.groupby("post_type")["likes"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Likes by Post Type:")
display(avg_likes_post_type)
```

```python
country_impressions = (
    df.groupby("country")["impression_count"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Impressions by Country:")
display(country_impressions)
```

```python
df["date"] = df["posted_at"].dt.date
df["year"] = df["posted_at"].dt.year
df["month"] = df["posted_at"].dt.month
df["month_name"] = df["posted_at"].dt.month_name()
df["day_name"] = df["posted_at"].dt.day_name()

display(
    df[
        [
            "posted_at",
            "date",
            "year",
            "month",
            "month_name",
            "day_name"
        ]
    ].head()
)
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 19, 29, 39, 49, 59, 100],
    labels=[
        "Teen",
        "20s",
        "30s",
        "40s",
        "50s",
        "60+"
    ]
)

print("Age Group Distribution:")

display(
    df["age_group"]
    .value_counts()
    .sort_index()
)
```

```python
df["log_likes"] = np.log1p(df["likes"])
df["log_comments"] = np.log1p(df["comments"])
df["log_shares"] = np.log1p(df["shares"])

display(
    df[
        [
            "likes",
            "log_likes",
            "comments",
            "log_comments",
            "shares",
            "log_shares"
        ]
    ].head()
)
```

```python
verified_posts = df[
    df["is_verified"] == True
].copy()

non_verified_posts = df[
    df["is_verified"] == False
].copy()

combined_df = pd.concat(
    [
        verified_posts,
        non_verified_posts
    ],
    ignore_index=True
)

print(
    "Original DataFrame:",
    df.shape
)

print(
    "Combined DataFrame:",
    combined_df.shape
)
```

```python
post_type_summary = (
    df.groupby("post_type")
    .agg(
        Average_Likes=("likes", "mean"),
        Average_Comments=("comments", "mean"),
        Average_Shares=("shares", "mean"),
        Average_Engagement_Rate=("engagement_rate", "mean"),
        Average_Impressions=("impression_count", "mean")
    )
    .round(2)
    .sort_values(
        "Average_Engagement_Rate",
        ascending=False
    )
)

display(post_type_summary)
```

```python
category_summary = (
    df.groupby("post_category")
    .agg(
        Average_Likes=("likes", "mean"),
        Average_Comments=("comments", "mean"),
        Average_Shares=("shares", "mean"),
        Average_Engagement=("engagement_score", "mean"),
        Average_Engagement_Rate=("engagement_rate", "mean")
    )
    .round(2)
    .sort_values(
        "Average_Engagement_Rate",
        ascending=False
    )
)

display(category_summary)
```

```python
sentiment_summary = (
    df.groupby("sentiment")
    .agg(
        Posts=("post_id", "count"),
        Average_Likes=("likes", "mean"),
        Average_Comments=("comments", "mean"),
        Average_Shares=("shares", "mean"),
        Average_Engagement_Rate=("engagement_rate", "mean")
    )
    .round(2)
    .sort_values(
        "Average_Engagement_Rate",
        ascending=False
    )
)

display(sentiment_summary)
```

```python
stat_columns = [
    "likes",
    "comments",
    "shares",
    "watch_time_sec",
    "engagement_rate",
    "follower_count"
]

statistics = pd.DataFrame(
    index=stat_columns
)

statistics["Mean"] = df[stat_columns].mean()

statistics["Median"] = df[stat_columns].median()

statistics["Mode"] = [
    df[column].mode()[0]
    for column in stat_columns
]

statistics["Standard Deviation"] = (
    df[stat_columns].std()
)
statistics["Variance"] = (
    df[stat_columns].var()
)

statistics["25th Percentile"] = (
    df[stat_columns].quantile(0.25)
)

statistics["50th Percentile"] = (
    df[stat_columns].quantile(0.50)
)

statistics["75th Percentile"] = (
    df[stat_columns].quantile(0.75)
)

statistics["Skewness"] = (
    df[stat_columns].skew()
)
display(
    statistics.round(3)
)
```

```python
monthly_engagement = (
    df.groupby(
        [
            "year",
            "month"
        ],
        as_index=False
    )["engagement_score"]
    .mean()
)

monthly_engagement["period"] = (
    monthly_engagement["year"]
    .astype(str)
    + "-"
    + monthly_engagement["month"]
    .astype(str)
    .str.zfill(2)
)

fig = px.line(
    monthly_engagement,
    x="period",
    y="engagement_score",
    markers=True,
    title=
    "Monthly Average Engagement Trend"
)

fig.update_xaxes(
    tickangle=45
)

fig.show()
```

```python
#Final Insights should include the following analysis:
# Content Performance
post_type_performance = (
    df.groupby("post_type")
    ["engagement_rate"]
    .mean()
    .sort_values(
        ascending=False
    )
)

display(
    post_type_performance
)

print(
    "Highest Engagement Post Type:",
    post_type_performance.index[0]
)
```

```python
category_performance = (
    df.groupby("post_category")
    ["engagement_rate"]
    .mean()
    .sort_values(
        ascending=False
    )
)

display(
    category_performance
)

print(
    "Best Performing Category:",
    category_performance.index[0]
)
```

```python
country_performance = (
    df.groupby("country")
    ["engagement_rate"]
    .mean()
    .sort_values(
        ascending=False
    )
)

display(
    country_performance
)
```

```python
# User Trends
age_engagement = (
    df.groupby(
        "age_group",
        observed=False
    )["engagement_rate"]
    .mean()
    .sort_values(
        ascending=False
    )
)

display(age_engagement)
```

```python
verified_analysis = (
    df.groupby("is_verified")
    .agg(
        Average_Engagement_Rate=
        ("engagement_rate", "mean"),

        Average_Likes=
        ("likes", "mean"),

        Average_Impressions=
        ("impression_count", "mean"),

        Number_of_Posts=
        ("post_id", "count")
    )
    .round(3)
)

display(
    verified_analysis
)
```

```python
#Behavioral Insights
day_impressions = (
    df.groupby("day_name")
    ["impression_count"]
    .mean()
)

day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_impressions = (
    day_impressions
    .reindex(day_order)
)
display(
    day_impressions
)

print(
    "\nBest Day for Average Impressions:"
)

print(
    day_impressions.idxmax()
)
```

```python
device_watch_time = (
    df.groupby("device_type")
    ["watch_time_sec"]
    .mean()
    .sort_values(
        ascending=False
    )
)

display(
    device_watch_time
)
```

```python
#Sentiment Analysis
sentiment_performance = (
    df.groupby("sentiment")
    .agg(
        Average_Engagement=
        ("engagement_rate", "mean"),

        Average_Likes=
        ("likes", "mean"),

        Average_Comments=
        ("comments", "mean"),

        Average_Shares=
        ("shares", "mean"),

        Post_Count=
        ("post_id", "count")
    )
    .round(3)
    .sort_values(
        "Average_Engagement",
        ascending=False
    )
)
display(
    sentiment_performance
)
```
